3.dv_Skills: Categorical skill rating derived from DV scores:
             "Very Good" if DV ≥ 18

```python
df['S.NO'] = range(1, len(df) + 1)
df
```

```python
df['DV'] = pd.to_numeric(df['DV'], errors='coerce')
def dv_skills(pp):
    if pp >= 18:
        return "Very Good"
    elif 12 <= pp <= 17:
        return "Good"
    else:
        return "Poor"
    
df['dv_Skills'] = df['DV'].apply(dv_skills)

print(df)
```

```python
df["Prog Skills"]=None
df
```

```python
value_count_DV = df['Prog Skills'].value_counts()
print("Value count for Prog Skills:\n", value_count_DV)
```

```python
value_count_DV = df['BEEE'].value_counts()
print("Value count for BEEE:\n", value_count_DV)
```

```python
df['PP'] = pd.to_numeric(df['DV'], errors='coerce')
def analytical_skills(dv):
    if dv >= 18:
        return "Very Good"
    elif 12 <= dv <= 17:
        return "Good"
    else:
        return "Poor"
    
df['Analytical Skills'] = df['DV'].apply(analytical_skills)

print(df)
```

```python
skill_counts = df['Analytical Skills'].value_counts()
plt.figure(figsize=(8, 8))
sns.set_theme(style="whitegrid")
colors = sns.color_palette('pastel')[0:3]
plt.pie(skill_counts, labels=skill_counts.index, autopct='%1.1f%%', 
        startangle=90, colors=colors, textprops={'fontsize': 12})
plt.title('Distribution of Analytical Skills', fontsize=16)
plt.show()
```

This pie chart shows how students are distributed across different levels of analytical skills, likely based on aggregated performance in specific subjects.

```python
sns.lineplot(x="PP", y="BEEE", data=df)
plt.xlabel("PP")
plt.ylabel("BEEE")
plt.title("Mid Marks of PP and BEEE")
plt.show()
```

This line plot with a shaded confidence interval shows the relationship between scores in PP and BEEE (Basic Electrical and Electronics Engineering).

```python
sns.barplot(x="PP", y="DV", data=df)
plt.xlabel("PP")
plt.ylabel("DV")
plt.title("Mid Marks: PP vs DV")
plt.show()
```

This chart shows the relationship between PP (presumably a subject like Programming Principles) and DV (possibly Data Visualization or another subject).
