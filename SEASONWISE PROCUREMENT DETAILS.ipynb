```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings 
warnings.filterwarnings("ignore")
```

Dataset Description: Season-wise Procurement Details 2023

This dataset contains information on the procurement of various agricultural commodities across different districts in multiple seasons. It includes details on the quantity procured, the number of farmers involved, and the total amount paid for the procurement.
1.District: The name of the district where the procurement took place. This helps in identifying regional variations in procurement.
2.Commodity: The type of agricultural product procured by the government or organization. Different commodities are procured based on season and demand.
3.Season: The season and year in which the procurement occurred. Agriculture follows a seasonal pattern, mainly Kharif (monsoon crops) and Rabi (winter crops).
4.Qty(MTs): The total quantity of the commodity procured, measured in Metric Tons (MTs). This represents the scale of procurement for each crop in each season.
5.No of Farmers: The number of farmers who participated in the procurement process. It indicates the reach of procurement efforts and farmer involvement.
6.Amount(Rs): The total amount (in Indian Rupees) paid for the procured commodity. This gives an insight into the financial impact of procurement for each district, season, and commodity.

```python
df['District'] = df['District'].str.strip().str.upper()
df['Commodity'] = df['Commodity'].str.strip().str.title()
df['Season'] = df['Season'].str.strip()
df
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['Qty(MTs)'], bins=20, kde=True, color="blue")
plt.xlabel("Quantity (MTs)")
plt.ylabel("Frequency")
plt.title("Distribution of Procurement Quantities")
plt.show()
```

This shows the distribution of procurement quantities

```python
top_commodities = df.groupby('Commodity')['Amount(Rs)'].sum().nlargest(10)
plt.figure(figsize=(10, 5))
sns.barplot(x=top_commodities.values, y=top_commodities.index, palette="viridis")
plt.xlabel("Total Procurement Amount (Rs)")
plt.ylabel("Commodity")
plt.title("Top 10 Commodities by Procurement Amount")
plt.show()
```

This shows the top 10 commodities by procurement amount

```python
district_data = df.groupby('District')['Amount(Rs)'].sum()
top_districts = district_data.nlargest(5)
plt.figure(figsize=(8, 8))
plt.pie(top_districts, labels=top_districts.index, autopct='%1.1f%%', colors=['#ffb3e6','#c2c2f0','#ff6666','#99ff99','#66b3ff'])
plt.title("Top 5 Districts by Procurement Amount")
plt.show()
```

This shows which districts contribute the most to procurement

```python
top_districts = df.groupby('District')['No of Farmers'].sum().nlargest(10)
plt.figure(figsize=(10, 5))
sns.barplot(x=top_districts.values, y=top_districts.index, palette="coolwarm")
plt.xlabel("Number of Farmers")
plt.ylabel("District")
plt.title("Top 10 Districts by Farmers Participation")
plt.show()
```

This shows which district have highest farmer participation

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x=df['Amount(Rs)'], color="green")
plt.xlabel("Procurement Amount (Rs)")
plt.title("Box Plot of Procurement Amount")
plt.show()
```

```python
seasonal_data = df.groupby('Season')['Qty(MTs)'].sum().reset_index()
plt.figure(figsize=(12, 5))
sns.lineplot(x='Season', y='Qty(MTs)', data=seasonal_data, marker="o", color="red")
plt.xticks(rotation=45)
plt.xlabel("Season")
plt.ylabel("Total Quantity (MTs)")
plt.title("Procurement Trends Over Different Seasons")
plt.show()
```

This shows the procurement over different seasons

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x=df['Qty(MTs)'], y=df['Amount(Rs)'], alpha=0.7, color="purple")
plt.xlabel("Quantity (MTs)")
plt.ylabel("Procurement Amount (Rs)")
plt.title("Relationship Between Quantity and Procurement Amount")
plt.show()
```

This shows the relationship between quantity and procurement amount

```python
commodity_data = df.groupby('Commodity')['Amount(Rs)'].sum()
top_commodities = commodity_data.nlargest(5)
plt.figure(figsize=(8, 8))
plt.pie(top_commodities, labels=top_commodities.index, autopct='%1.1f%%', colors=['#ff9999','#66b3ff','#99ff99','#ffcc99','#c2c2f0'])
plt.title("Top 5 Commodities by Procurement Amount")
plt.show()
```

This shows which commodities have the highest procurement

```python
g = sns.FacetGrid(df, col="Season", col_wrap=3, height=4, sharex=False, sharey=False)
g.map(sns.kdeplot, "Qty(MTs)", fill=True, color="blue")
g.fig.suptitle("FacetGrid KDE Plot of Procurement Quantity by Season", fontsize=14)
plt.tight_layout()
plt.show()
```

This shows the procurement quantity across different seasons

District-Wise Insights:

Financial Impact:

This dataset gives valuable insights into seasonal crop production, farmer involvement, and the scale of procurement. It helps in understanding how agricultural policies and market support are working for farmers during different seasons.
